### What is an embedding?

An embedding turns text into a list of numbers — a vector — that represents its meaning. Texts with similar meaning tend to produce vectors that are closer together.

```text
"physicist who developed the theory of relativity"
                    ↓
            Qwen3-Embedding-8B
                    ↓
          [0.12, -0.37, ...]
                    ↓
       compare with 100,000 vectors
                    ↓
             Albert Einstein
```

That is the key idea behind **semantic search**: retrieve by meaning, not only by exact words.

## 4. English semantic search — search by meaning, not exact wording / Tìm kiếm ngữ nghĩa tiếng Anh — tìm theo ý nghĩa

**English:** This is the first visible semantic-search demonstration. Each query is embedded live, compared against the 100K-vector index, and the nearest results are returned.

### Example to notice
For a query such as:

> **physicist who developed the theory of relativity**

the query does **not** contain the words `Albert Einstein`, yet a good semantic embedding model should still rank **Albert Einstein** at or near the top because the description matches his meaning.

Likewise:

> **former capital of West Germany**

can retrieve **Bonn** even though the user never typed the word `Bonn`.

### How to read the table
- `rank=1` is the best match.
- `score` is semantic similarity in the embedding space; higher means closer for this query.
- `English` / `Tiếng Việt` show the bilingual Wikidata labels.

**Tiếng Việt:** Đây là phần demo semantic search đầu tiên mà người xem có thể quan sát trực tiếp. Mỗi query được encode live, so sánh với index 100K vector và trả về các kết quả gần nhất về mặt ngữ nghĩa.

### Ví dụ đáng chú ý
Với query:

> **physicist who developed the theory of relativity**

câu hỏi **không chứa** chữ `Albert Einstein`, nhưng model embedding tốt vẫn có thể xếp **Albert Einstein** ở vị trí đầu hoặc gần đầu vì mô tả trong query phù hợp với ý nghĩa liên quan tới Einstein.

Tương tự:

> **former capital of West Germany**

có thể trả về **Bonn** dù người dùng không hề gõ từ `Bonn`.

### Cách đọc bảng kết quả
- `rank=1` là kết quả phù hợp nhất.
- `score` biểu diễn mức độ gần về ngữ nghĩa trong embedding space; score cao hơn nghĩa là gần hơn đối với query này.
- `English` / `Tiếng Việt` là nhãn Wikidata song ngữ.

Điểm cốt lõi: người dùng có thể **mô tả thứ họ muốn tìm** thay vì phải biết chính xác từ khóa hoặc tên thực thể đã được lưu trong dữ liệu.


````python
from IPython.display import display
import pandas as pd

def search(q,k=5):
    r=httpx.post(BASE+'/v1/search',headers=HEADERS,json={'query':q,'top_k':k},timeout=120)
    r.raise_for_status()
    return r.json()['data']

def show(title,q):
    hits=search(q,5)
    print()
    print(title)
    print('Query:',q)
    display(pd.DataFrame([{'rank':x['rank'],'score':round(x['score'],4),'qid':x.get('qid'),'English':x.get('label_en'),'Tiếng Việt':x.get('label_vi')} for x in hits]))

EN_EXAMPLES=[('History / Lịch sử','former capital of West Germany'),('Person / Nhân vật','physicist who developed the theory of relativity'),('Technical / Kỹ thuật','Java programming language')]
for t,q in EN_EXAMPLES:
    show(t,q)
````

## 6. Cross-lingual retrieval — different languages, same semantic neighborhood / Retrieval xuyên ngôn ngữ

**English:** Here we ask equivalent questions in **English and Vietnamese** and compare the retrieved Wikidata entities.

### The intuition
If the model truly understands both languages semantically, equivalent EN and VI queries should land near the same region of vector space and therefore retrieve many of the same entities.

```text
English query ─┐
               ├─→ similar semantic region ─→ similar top results
Vietnamese ────┘
```

### How to read the output
- The top entity is compared for each EN/VI pair.
- `top-10 overlap` tells us how many of the ten best results are shared between the two languages.
- A high overlap means the language changed, but the semantic neighborhood stayed similar.

**Tiếng Việt:** Ở đây chúng ta đặt các câu hỏi tương đương bằng **tiếng Anh và tiếng Việt**, sau đó so sánh các thực thể Wikidata được retrieval.

### Trực giác phía sau
Nếu model thật sự hiểu hai ngôn ngữ ở mức ngữ nghĩa, hai query EN và VI tương đương nên rơi vào các vùng gần nhau trong vector space và vì vậy trả về nhiều thực thể giống nhau.

```text
Query tiếng Anh ─┐
                 ├─→ vùng ngữ nghĩa gần nhau ─→ top results tương tự
Query tiếng Việt ─┘
```

### Cách đọc output
- So sánh top entity của từng cặp EN/VI.
- `top-10 overlap` cho biết có bao nhiêu kết quả trong top 10 xuất hiện ở cả hai ngôn ngữ.
- Overlap cao nghĩa là ngôn ngữ đã thay đổi nhưng semantic neighborhood vẫn tương tự.

Đây là nền tảng cho cross-language search: người dùng có thể hỏi bằng tiếng Việt nhưng vẫn tìm được nội dung được biểu diễn hoặc gắn nhãn bằng tiếng Anh, và ngược lại.

### Reading the expanded table / Cách đọc bảng mở rộng
The canonical Wikidata **QID columns are intentionally preserved** for auditability. Human-readable English and Vietnamese entity labels are shown beside them so non-specialists can understand the result immediately without losing the stable identifier.

Các cột **QID canonical vẫn được giữ nguyên** để phục vụ kiểm tra/audit. Tên thực thể tiếng Anh và tiếng Việt được bổ sung bên cạnh để người không quen Wikidata vẫn hiểu ngay kết quả mà không mất định danh ổn định.


````python
PAIRS=[('Microsoft','Microsoft technology company','công ty Microsoft'),('Marie Curie','Marie Curie scientist','nhà khoa học Marie Curie'),('Computer science','computer science','khoa học máy tính')]
for name,enq,viq in PAIRS:
    a=search(enq,10); b=search(viq,10)
    ai=[x.get('qid') for x in a]; bi=[x.get('qid') for x in b]
    overlap=len(set(ai)&set(bi))/10
    print(f'\n{name}: top-10 overlap={overlap:.0%}')
    display(pd.DataFrame({
        'EN top QID':ai[:5],
        'EN entity':[x.get('label_en') for x in a[:5]],
        'EN entity / Tiếng Việt':[x.get('label_vi') for x in a[:5]],
        'VI top QID':bi[:5],
        'VI entity':[x.get('label_en') for x in b[:5]],
        'VI entity / Tiếng Việt':[x.get('label_vi') for x in b[:5]],
    }))
````

## 7. Instruction-aware retrieval — show the model what role the text plays / Retrieval có instruction

**English:** Qwen3 Embedding can distinguish between text used as a **query** and ordinary text/document content. This section compares the same text with and without retrieval-oriented query instruction.

### Why this is impressive
A short query such as:

> **Einstein**

contains almost no context. Without query-oriented handling, the expected entity may not appear in the top results. With the retrieval instruction enabled, the expected entity can move dramatically upward — in the validated rehearsal, short `Einstein` and Vietnamese `công ty Microsoft` both moved from **not in Top 5** to **rank 1**.

### How to read the table
- `without instruction`: ranking when the text is treated like ordinary content.
- `with instruction`: ranking when the model is explicitly told the text is a retrieval query.
- The expected QID should move higher when the instruction helps.

**Tiếng Việt:** Qwen3 Embedding có thể phân biệt text đang đóng vai trò **query tìm kiếm** hay chỉ là text/document thông thường. Phần này encode cùng một nội dung theo hai cách: có và không có retrieval-oriented query instruction.

### Vì sao kết quả này đáng chú ý
Một query ngắn như:

> **Einstein**

hầu như không có ngữ cảnh. Nếu không xử lý theo vai trò query, thực thể mong đợi có thể không xuất hiện trong top results. Khi bật retrieval instruction, thứ hạng có thể tăng rất mạnh — trong lần rehearsal đã xác minh, `Einstein` và query tiếng Việt `công ty Microsoft` đều đi từ **không có trong Top 5** lên **rank 1**.

### Cách đọc bảng
- `without instruction`: ranking khi text được xử lý như nội dung thông thường.
- `with instruction`: ranking khi model được nói rõ đây là một retrieval query.
- QID mong đợi nên đi lên thứ hạng cao hơn khi instruction thực sự hữu ích.

Điều này cho thấy embedding không chỉ phụ thuộc vào nội dung câu chữ mà còn có thể tận dụng **vai trò/nhiệm vụ** của text để tạo representation phù hợp hơn cho retrieval.

### Reading the expanded table / Cách đọc bảng mở rộng
The table keeps the exact Wikidata **QID** for every result and adds English/Vietnamese labels next to it. This makes the before/after ranking understandable to humans while preserving the canonical machine-readable identity.

Bảng vẫn giữ chính xác **QID Wikidata** của từng kết quả, đồng thời thêm nhãn tiếng Anh/tiếng Việt. Nhờ vậy người xem có thể hiểu ngay thay đổi thứ hạng trước/sau instruction mà vẫn giữ định danh machine-readable canonical.


````python
import numpy as np
from app.search.index import IndexMetadata, VectorIndex
meta=IndexMetadata(**json.loads((INDEX_DIR/'index-metadata.json').read_text())); base_index=VectorIndex.load(INDEX_DIR,expected_metadata=meta); ab_index=base_index.project(1024)
def embed(text,dim=1024,is_query=True):
    r=httpx.post(BASE+'/v1/embeddings',headers=HEADERS,json={'input':text,'dimensions':dim,'is_query':is_query},timeout=120); r.raise_for_status(); return np.asarray(r.json()['data'][0]['embedding'],dtype=np.float32)
def ranks(q,iq): return ab_index.search(embed(q,1024,iq),5)
AB=[('Einstein descriptive','physicist who developed the theory of relativity','Q937'),('Einstein short','Einstein','Q937'),('Microsoft VI','công ty Microsoft','Q2283')]
for name,q,expected in AB:
    plain=ranks(q,False); inst=ranks(q,True)
    def pos(h):
        ids=[x.row.get('qid') for x in h]; return ids.index(expected)+1 if expected in ids else 'not in top 5'
    def cols(h,prefix):
        return {
            f'{prefix} QID':[x.row.get('qid') for x in h],
            f'{prefix} entity':[x.row.get('label_en') for x in h],
            f'{prefix} entity / Tiếng Việt':[x.row.get('label_vi') for x in h],
        }
    print(f'\n{name} | expected={expected} | without={pos(plain)} | with={pos(inst)}')
    data={}; data.update(cols(plain,'without instruction')); data.update(cols(inst,'with instruction'))
    display(pd.DataFrame(data))
````

## 8. Semantic similarity — turn meaning into a measurable distance / Đo độ tương đồng ngữ nghĩa

**English:** Embeddings are useful beyond search. We can directly compare two vectors with **cosine similarity**.

A simple interpretation:

```text
closer to 1.0  → meanings are very similar
closer to 0.0  → meanings are weakly related
```

We compare three intuitive cases:
1. an English sentence and its Vietnamese translation;
2. two English paraphrases;
3. two unrelated sentences.

### Why this matters
The same mechanism can support duplicate detection, clustering, recommendations, matching, and many other semantic applications.

**Tiếng Việt:** Embedding không chỉ hữu ích cho search. Ta có thể so sánh trực tiếp hai vector bằng **cosine similarity** để đo mức độ gần nhau về ngữ nghĩa.

Cách hiểu đơn giản:

```text
gần 1.0  → ý nghĩa rất giống nhau
gần 0.0  → ít liên quan về mặt ngữ nghĩa
```

Notebook so sánh ba trường hợp dễ hiểu:
1. một câu tiếng Anh và bản dịch tiếng Việt của nó;
2. hai câu tiếng Anh diễn đạt lại cùng một ý;
3. hai câu không liên quan.

### Vì sao bước này quan trọng
Cùng cơ chế này có thể dùng cho phát hiện nội dung trùng/na ná nhau, clustering, recommendation, matching và nhiều bài toán semantic khác.

Khi một cặp dịch EN/VI hoặc hai câu paraphrase có similarity cao, còn hai câu không liên quan có similarity thấp, người xem có thể trực tiếp thấy vector đang phản ánh **ý nghĩa**, chứ không chỉ sự trùng khớp ký tự.


````python
def cosine(a,b): return float(np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b)))
SIM=[('Translation / Bản dịch','A cat is sleeping on the sofa.','Một con mèo đang ngủ trên ghế sofa.'),('Paraphrase','The meeting was postponed until Monday.','The meeting has been moved to Monday.'),('Unrelated / Không liên quan','A cat is sleeping on the sofa.','Thị trường chứng khoán tăng điểm hôm nay.')]
rows=[]
for name,a,b in SIM: rows.append({'example':name,'cosine_similarity':round(cosine(embed(a,1024,False),embed(b,1024,False)),4),'text A':a,'text B':b})
display(pd.DataFrame(rows))
````

## 9. Flexible embedding dimensions — choose a deployment trade-off / Kích thước embedding linh hoạt

**English:** The full model can return a **4096-dimensional** embedding, but Qwen3 also supports smaller representations such as **1024** or **256** dimensions.

Why would anyone use a smaller vector?

```text
4096 dims → richest representation, largest storage / compute footprint
1024 dims → smaller representation
 256 dims → much more compact
```

A smaller vector can reduce index storage and comparison cost. The right choice depends on the application and its accuracy/efficiency trade-off.

### What this cell proves
For English, Vietnamese, and bilingual text, we request 256, 1024, and 4096 dimensions and verify that the API returns exactly the requested size.

`norm` being close to `1.0` shows the returned vectors are normalized as expected.

**Tiếng Việt:** Model đầy đủ có thể trả về embedding **4096 chiều**, nhưng Qwen3 cũng hỗ trợ representation nhỏ hơn như **1024** hoặc **256** chiều.

Vì sao lại muốn dùng vector nhỏ hơn?

```text
4096 chiều → representation đầy đủ nhất, tốn storage/compute hơn
1024 chiều → nhỏ hơn đáng kể
 256 chiều → rất gọn, phù hợp khi ưu tiên tài nguyên
```

Vector nhỏ hơn có thể giảm dung lượng index và chi phí so sánh. Lựa chọn đúng phụ thuộc vào trade-off giữa chất lượng retrieval và tài nguyên của ứng dụng.

### Cell này chứng minh điều gì
Với text tiếng Anh, tiếng Việt và song ngữ, notebook yêu cầu lần lượt 256, 1024 và 4096 chiều rồi xác minh API thực sự trả đúng kích thước được yêu cầu.

`norm` gần `1.0` cho thấy vector trả về đã được normalize như mong đợi.

Điểm thực tế ở đây là cùng một model có thể phục vụ nhiều kiểu deployment khác nhau: từ hệ thống cần representation đầy đủ đến ứng dụng muốn giảm memory/storage và chi phí tính toán.


````python
MRL_TEXTS=[('English','Semantic search connects meaning, not only keywords.'),('Tiếng Việt','Tìm kiếm ngữ nghĩa dựa trên ý nghĩa thay vì chỉ từ khóa.'),('Bilingual','Qwen embeddings hỗ trợ semantic retrieval across languages.')]
rows=[]
for kind,text in MRL_TEXTS:
    for dim in (256,1024,4096):
        t=time.perf_counter(); v=embed(text,dim,False); rows.append({'example':kind,'requested_dim':dim,'actual_dim':len(v),'norm':round(float(np.linalg.norm(v)),4),'latency_s':round(time.perf_counter()-t,3)})
display(pd.DataFrame(rows))
````

## 10. Dual-T4 concurrency — prove both GPUs are serving requests / Chứng minh cả hai T4 đều phục vụ request

**English:** A production service rarely receives only one request at a time. This section sends **1, 8, and 16 live embedding requests** and then checks worker metrics.

### What success means
```text
16 concurrent requests
16 successful
0 failed
GPU worker 0 active ✅
GPU worker 1 active ✅
```

The exact request count per worker does not need to be perfectly balanced. The important point is that **both workers actually process work**.

This is a concurrency demonstration, not a claim that two GPUs are always exactly 2× faster than one GPU.

**Tiếng Việt:** Một production service hiếm khi chỉ nhận đúng một request tại một thời điểm. Phần này gửi **1, 8 và 16 embedding request live** rồi đọc worker metrics.

### Thành công nghĩa là gì
```text
16 request đồng thời
16 thành công
0 thất bại
GPU worker 0 có xử lý việc ✅
GPU worker 1 có xử lý việc ✅
```

Số request trên mỗi worker không cần cân bằng tuyệt đối. Điều quan trọng là **cả hai worker đều thực sự xử lý workload**, chứ không phải chỉ một GPU hoạt động còn GPU kia đứng yên.

Đây là demo về concurrency và serving architecture, **không phải** tuyên bố rằng hai GPU luôn nhanh chính xác gấp 2 lần một GPU. Tốc độ thực tế còn phụ thuộc batch shape, scheduler, overhead, sequence length và nhiều yếu tố khác.


````python
from concurrent.futures import ThreadPoolExecutor
def one(i):
    r=httpx.post(BASE+'/v1/embeddings',headers=HEADERS,json={'input':f'quick demo request {i}','dimensions':1024,'is_query':True},timeout=120); return r.status_code
for n in (1,8,16):
    t=time.perf_counter();
    with ThreadPoolExecutor(max_workers=min(n,8)) as ex: codes=list(ex.map(one,range(n)))
    print(f'workload={n} successful={sum(c==200 for c in codes)}/{n} elapsed={time.perf_counter()-t:.2f}s')
metrics=httpx.get(BASE+'/metrics',headers=HEADERS,timeout=10).json(); print('items_per_worker=',metrics.get('items_per_worker')); assert '0' in metrics.get('items_per_worker',{}) and '1' in metrics.get('items_per_worker',{}); print('DUAL_T4_SHOWCASE=PASS')
````
